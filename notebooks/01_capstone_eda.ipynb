```python
df["failure"].value_counts(normalize=True).rename("share")
```

## Feature distributions by outcome

Machines that eventually fail should show higher temperature, higher vibration, larger pressure deviation from the nominal band, and more accumulated operating hours.

```python
features = ["temperature", "vibration", "pressure", "operating_hours"]

fig, axes = plt.subplots(2, 2, figsize=(11, 8))
for ax, feature in zip(axes.ravel(), features):
    for label, group in df.groupby("failure"):
        ax.hist(group[feature], bins=30, alpha=0.5, label=f"failure={label}")
    ax.set_title(feature)
    ax.legend()
fig.tight_layout()
```

```python
df[features + ["failure"]].corr()["failure"].sort_values(ascending=False)
```
